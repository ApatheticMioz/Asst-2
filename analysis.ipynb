## Task 2 Data Prep (Gapminder)

```python
base_path = 'Data/ddf--gapminder--fasttrack-master/countries_etc_datapoints/'
geo_path = 'Data/ddf--gapminder--fasttrack-master/'

print("Loading GDP...")
gdp_df = pd.read_csv(base_path + 'ddf--datapoints--gdp_pcap--by--country--time.csv').rename(columns={'gdp_pcap': 'income'})

print("Loading Life Expectancy...")
lex_df = pd.read_csv(base_path + 'ddf--datapoints--lex--by--country--time.csv').rename(columns={'lex': 'life_exp'})

print("Loading Population...")
pop_df = pd.read_csv(base_path + 'ddf--datapoints--pop--by--country--time.csv').rename(columns={'pop': 'population'})

# 2. Load the Country Entities (to get Regions/Continents)
print("Loading Region Data...")
# Matches [ddf--entities--geo--country.csv] from your output
country_ent_df = pd.read_csv(geo_path + 'ddf--entities--geo--country.csv')

# We only need the code (country), the full name (name), and the region (world_4region)
region_df = country_ent_df[['country', 'name', 'world_4region']]

print("All files loaded successfully.")
```

```python
# 1. Merge the dataframes on 'country' and 'time'
# We use 'inner' join to ensure we only keep records that contain all three metrics
merged_df = pd.merge(gdp_df, lex_df, on=['country', 'time'], how='inner')
merged_df = pd.merge(merged_df, pop_df, on=['country', 'time'], how='inner')

# 2. Merge with Region data to get the full Country Name and Continent
# This merge happens only on 'country'
merged_df = pd.merge(merged_df, region_df, on='country', how='inner')

# 3. Filter for years 1800 to 2020 
# (Assignment Requirement: "last two centuries")
merged_df = merged_df[(merged_df['time'] >= 1800) & (merged_df['time'] <= 2020)]

# 4. Handle Missing Values
# Drop rows where region is missing (sometimes small islands/territories don't have a region)
merged_df = merged_df.dropna(subset=['world_4region'])

print(f"Merged Data Shape: {merged_df.shape}")
merged_df.head()
```

### EDA and Cleaning

```python
import json

# 1. Clean up column names for the Javascript side
final_df = merged_df.rename(columns={
    'time': 'year', 
    'world_4region': 'continent'
})

# 2. Convert to Hierarchical JSON: Array of Years -> Array of Countries
# This structure is highly optimized for D3.js animation loops
formatted_data = []
years = sorted(final_df['year'].unique())

for year in years:
    year_data = final_df[final_df['year'] == year]
    
    # Convert that year's data to a list of dictionaries
    countries_list = year_data[['country', 'name', 'continent', 'income', 'life_exp', 'population']].to_dict('records')
    
    formatted_data.append({
        "year": int(year),
        "countries": countries_list
    })

# 3. Save to file
output_path = 'Data/gapminder_clean.json'
with open(output_path, 'w') as f:
    json.dump(formatted_data, f)

print(f"SUCCESS: Data saved to {output_path}")
print(f"Contains data for {len(formatted_data)} years.")
```
